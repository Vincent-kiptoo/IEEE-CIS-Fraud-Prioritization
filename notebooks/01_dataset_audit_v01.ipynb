### Feature Dictionary: train_transaction

The following table describes the features available in the IEEE-CIS
Fraud Detection transaction dataset, their data types, and their
documented meanings.

| Feature / Feature Group | Data Type | Feature Definition |
|:---|:---|:---|
| TransactionID | BIGINT | Unique identifier assigned to each transaction. |
| isFraud | BIGINT | Target variable indicating whether a transaction was fraudulent (1) or legitimate (0). |
| TransactionDT | BIGINT | Time delta from a reference timestamp. The actual timestamp is not provided. |
| TransactionAmt | DOUBLE | Transaction amount. |
| ProductCD | VARCHAR | Product code associated with the transaction. |
| card1 | BIGINT | Payment card information, field 1. |
| card2 | DOUBLE | Payment card information, field 2. |
| card3 | DOUBLE | Payment card information, field 3. |
| card4 | VARCHAR | Payment card network or type, such as Visa or Mastercard. |
| card5 | DOUBLE | Payment card information, field 5. |
| card6 | VARCHAR | Payment card category, such as credit or debit. |
| addr1 | DOUBLE | Address information, field 1. |
| addr2 | DOUBLE | Address information, field 2. |
| dist1 | DOUBLE | Distance-related transaction feature, field 1. |
| dist2 | DOUBLE | Distance-related transaction feature, field 2. |
| P_emaildomain | VARCHAR | Purchaser's email domain. |
| R_emaildomain | VARCHAR | Recipient's email domain. |
| C1–C14 | DOUBLE | Counting features representing the number of transactions or related records associated with a particular entity or condition. Exact individual definitions are anonymized. |
| D1–D15 | DOUBLE | Time-delta features representing elapsed time between transactions or related events. Exact individual definitions are anonymized. |
| M1–M3 | BOOLEAN | Match-related indicators describing whether certain transaction attributes match. |
| M4 | VARCHAR | Match-related feature containing categorical values. |
| M5–M9 | BOOLEAN | Additional match-related indicators. |
| V1–V339 | DOUBLE | Vesta-engineered anonymized features containing transaction-related identity, behavioural and other derived information. Individual definitions are not publicly disclosed. |

**Data type interpretation:**

- BIGINT: Large integer values.
- DOUBLE: Numeric values supporting decimal precision.
- VARCHAR: Variable-length character or categorical data.
- BOOLEAN: Binary TRUE/FALSE values.

**Important notes:**

1. `TransactionID` is an identifier, not a predictive measurement.
2. `isFraud` is the target variable for supervised fraud classification.
3. `TransactionDT` represents relative time rather than a conventional timestamp.
4. C, D, M and V feature families contain anonymized or engineered information.
5. Feature definitions should not be confused with their statistical distributions.

### 2: Training-test schema comparison

```python
%%sql

-- Compares train_transaction against train_identity to the training dataset (X_train)

SELECT *
    FROM train_transaction AS tt

LEFT JOIN train_identity AS ti
    ON tt.TransactionID = ti.TransactionID
```

```python
%%sql

-- Checks the row counts after merging the dataset
SELECT 
    COUNT(*) AS total_rows,
    COUNT(ti.TransactionID) AS matched_identity_rows,
    COUNT(*) - COUNT(ti.TransactionID) AS missing_identity_rows
FROM train_transaction AS tt
LEFT JOIN train_identity AS ti
    ON tt.TransactionID = ti.TransactionID;
```

```python
train_cols = %sql SELECT * FROM train_transaction AS tt LEFT JOIN train_identity AS ti ON tt.TransactionID = ti.TransactionID LIMIT 0

train_df = train_cols.DataFrame()

print(f"The total number of columns is: {len(train_df.columns)}")
```

```python
%%sql

-- Joins both test transaction and identity dataset on left join to form the testing dataset (X_test)

SELECT * 
FROM test_transaction AS tt
LEFT JOIN test_identity AS ti
    ON tt.TransactionID = ti.TransactionID
```

```python
%%sql

-- Compares the number of rows after merging the two datasets

SELECT
    COUNT(*) AS total_rows,
    COUNT(ti.TransactionID) AS matched_identity_rows,
    COUNT(*) - COUNT(ti.TransactionID) AS missing_identity_rows
FROM test_transaction AS tt
LEFT JOIN test_identity AS ti
    ON tt.TransactionID = ti.TransactionID
```

```python
test_df = %sql SELECT * FROM test_transaction AS tt LEFT JOIN test_identity AS ti ON tt.TransactionID = ti.TransactionID LIMIT 0

test_cols = test_df.DataFrame()

print(f"The total number of rows is: {len(test_cols.columns)}")
```
